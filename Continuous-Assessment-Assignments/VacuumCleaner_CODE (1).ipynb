**Vacuum Cleaner Agent Using Artificial Intelligence(Simple Reflex Agent)**

**Aim :** To design and implement a simple AI-based college vacuum cleaner agent that can move between different college locations, detect dirty areas, clean them according to their priority, and manage its battery while performing cleaning tasks.

* To implement a vacuum cleaner as an intelligent agent.

**Concept :**
An AI agent is a system that observes its environment, makes decisions, and performs actions to achieve a particular goal.

In this experiment:

Environment  →  College

1.  Agent        →  Vacuum Cleaner
2.  Goal         →  Clean dirty college locations
3.  Sensor       →  Detect dirt level
4.  GAction       →  Move and clean
5.  Performance  →  Cleaning score

```python
import random

#Create Environment
class CollegeEnvironment:
    def __init__(self):
        self.rooms = {
            "Classroom": {
                "dirt": random.randint(20, 90),
                "priority": 3,
                "occupied": random.choice([True, False])
            },

            "Laboratory": {
                "dirt": random.randint(20, 90),
                "priority": 5,
                "occupied": random.choice([True, False])
            },

            "Library": {
                "dirt": random.randint(20, 90),
                "priority": 2,
                "occupied": random.choice([True, False])
            },

            "Cafeteria": {
                "dirt": random.randint(20, 90),
                "priority": 4,
                "occupied": random.choice([True, False])
            },

            "Corridor": {
                "dirt": random.randint(20, 90),
                "priority": 3,
                "occupied": False
            }
        }

#College Vacuum Agent Class
class CollegeVacuumAgent:
    def __init__(self, environment):
        self.environment = environment
        self.currentLocation = "Corridor"
        self.battery = 100
        self.score = 0

#display environmnet before clean
    def display_environment(self):
        print("\nCollege Environment")
        print("----------------------------")

        for room, data in self.environment.rooms.items():
            print(
                room,
                " --> Dirt: ", data["dirt"],
                " , Priority: ", data["priority"],
                " , Occupied: ", data["occupied"]
            )
#claculat Performance score
    def calculate_score(self, room):
        data = self.environment.rooms[room]

        if data["occupied"]:
            return -1

        return data["dirt"] + (data["priority"] * 10)

    def select_room(self):
        best_room = None
        best_score = -1

        for room in self.environment.rooms:
            if self.environment.rooms[room]["dirt"] > 0:
                room_score = self.calculate_score(room)

                if room_score > best_score:
                    best_score = room_score
                    best_room = room

        return best_room

    def move_to(self, room):
        print("\nMoving from", self.currentLocation, "to", room)

        self.currentLocation = room
        self.battery -= 10
        self.score -= 1

    def clean_room(self, room):
        data = self.environment.rooms[room]

        print("Vacuum is checking", room)

        if data["occupied"]:
            print(room, "is occupied. Skipping...")
            return

        if data["dirt"] > 0:
            print(room, "is Dirty")
            print("Dirt Level:", data["dirt"])

            data["dirt"] = 0
            self.battery -= 15
            self.score += 10

            print(room, "has been Cleaned")

        else:
            print(room, "is Already Clean")


    def recharge(self):
        print("\nBattery is Low")
        print("Vacuum is going to Charging Station")

        self.currentLocation = "Charging Station"
        self.battery = 100

        print("Battery Charged to 100%")

    def clean_college(self):
        print("\nVacuum starts cleaning the college")

        while True:
            dirty_rooms = [
                room
                for room in self.environment.rooms
                if self.environment.rooms[room]["dirt"] > 0
            ]

            if not dirty_rooms:
                break

            if self.battery < 30:
                self.recharge()

            room = self.select_room()

            if room is None:
                break

            self.move_to(room)
            self.clean_room(room)

            print("Battery:", self.battery)
            print("Score:", self.score)

    def final_report(self):
        print("\nFinal College Environment")
        print("----------------------------")

        for room, data in self.environment.rooms.items():
            print(room, "| Dirt:", data["dirt"])

        print("\nFinal Location:", self.currentLocation)
        print("Final Battery:", self.battery)
        print("Performance Score:", self.score)


environment = CollegeEnvironment()

agent = CollegeVacuumAgent(environment)

agent.display_environment()

agent.clean_college()

agent.final_report()
```

**Conclusion :** The College Vacuum Cleaner Agent was successfully implemented using Python and basic artificial intelligence concepts.

The system models a college environment containing multiple locations with different dirt levels, priorities, and occupancy conditions. The vacuum cleaner observes the environment and selects a suitable location for cleaning instead of simply following a fixed sequence.

Thus, this experiment provides a simple foundation for developing more advanced autonomous cleaning systems for colleges, hospitals, offices, shopping malls, and other large environments.
